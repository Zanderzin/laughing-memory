```python
import pandas as pd
import plotly as plt
```

### Ler o DataFrame

```python
dados = pd.read_csv("C:\\Users\\T1092434\\Downloads\\dados_dashboard.csv", header=None, names=["Matricula", "Nome_Funcionario", "Avaliavel", "Data_Ultimo_Eps", "Situacao_Eps", "Dias_Para_Vencimento", "Status_Indicador", "Cargo", "Prefixo", "Dependencia", "Codigo_Uor", "Uor", "Prefixo_Ajure", "Ajure"])
```

```python
dados = dados.drop([0, 1]).reset_index(drop=True)
```

```python
print(dados.head())
```

## Separando prefixos

```python
prefixos = dados["Prefixo"].tolist()
prefixos = list(set(prefixos))
print(prefixos)
```

## Separando DFs

```python
# 1) Converta a coluna para datetime (considerando formato dia/mês/ano)
dados["Data_Ultimo_Eps"] = pd.to_datetime(dados["Data_Ultimo_Eps"], dayfirst=True, errors="coerce")

# 2) Defina a data limite
limite = pd.Timestamp("2025-06-30")

# 3) Filtre as linhas ANTES de 30/06/2025
dados_antes = dados[dados["Data_Ultimo_Eps"] < limite].copy()
print(dados_antes)

```

# Porcentagem

```python
porcentagem = len(dados_antes) / len(dados) * 100
print(f"A porcentagem de pessoas que precisam fazer o EPS é de: {round(porcentagem, 2)}%")
```

```python
# Contagem total por prefixo no DF inteiro
totais = dados["Prefixo"].value_counts()

# Contagem por prefixo apenas no subset "antes"
antes = dados_antes["Prefixo"].value_counts()

# Alinha pelos mesmos prefixos e calcula a % por prefixo
porc_por_prefixo = (antes / totais * 100).fillna(0).sort_index()

print(porc_por_prefixo)
```
